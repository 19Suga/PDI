```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("bonito.png")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

R = img[:,:,0]
G = img[:,:,1]
B = img[:,:,2]

mosaico = np.zeros((img.shape[0], img.shape[1]), dtype=np.uint8)
mosaico[0::2,0::2] = R[0::2,0::2]  
mosaico[0::2,1::2] = G[0::2,1::2]  
mosaico[1::2,0::2] = G[1::2,0::2]  
mosaico[1::2,1::2] = B[1::2,1::2]  

recon_bg = cv2.cvtColor(mosaico, cv2.COLOR_BAYER_BG2RGB)
recon_gb = cv2.cvtColor(mosaico, cv2.COLOR_BAYER_GB2RGB)
recon_rg = cv2.cvtColor(mosaico, cv2.COLOR_BAYER_RG2RGB)
recon_gr = cv2.cvtColor(mosaico, cv2.COLOR_BAYER_GR2RGB)

color_bg = cv2.applyColorMap(cv2.cvtColor(recon_bg, cv2.COLOR_RGB2GRAY), cv2.COLORMAP_JET)
color_gb = cv2.applyColorMap(cv2.cvtColor(recon_gb, cv2.COLOR_RGB2GRAY), cv2.COLORMAP_HOT)
color_rg = cv2.applyColorMap(cv2.cvtColor(recon_rg, cv2.COLOR_RGB2GRAY), cv2.COLORMAP_COOL)
color_gr = cv2.applyColorMap(cv2.cvtColor(recon_gr, cv2.COLOR_RGB2GRAY), cv2.COLORMAP_SPRING)

plt.figure(figsize=(12,8))

plt.subplot(231)
plt.imshow(img)
plt.title("Imagen Original")

plt.subplot(232)
plt.imshow(mosaico, cmap="gray")
plt.title("Mosaico Bayer")

plt.subplot(233)
plt.imshow(cv2.cvtColor(color_bg, cv2.COLOR_BGR2RGB))
plt.title("Bayer BG → RGB")

plt.subplot(234)
plt.imshow(cv2.cvtColor(color_gb, cv2.COLOR_BGR2RGB))
plt.title("Bayer GB → RGB")

plt.subplot(235)
plt.imshow(cv2.cvtColor(color_rg, cv2.COLOR_BGR2RGB))
plt.title("Bayer RG → RGB")

plt.subplot(236)
plt.imshow(cv2.cvtColor(color_gr, cv2.COLOR_BGR2RGB))
plt.title("Bayer GR → RGB")

plt.tight_layout()
plt.show()
```

rgb-yuv

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("bonito.png")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# RGB a YUV
yuv = cv2.cvtColor(img, cv2.COLOR_RGB2YUV)

# YUV a RGB
rgb_recuperado = cv2.cvtColor(yuv, cv2.COLOR_YUV2RGB)

plt.figure(figsize=(10,5))

plt.subplot(1,3,1)
plt.imshow(img)
plt.title("Imagen RGB")

plt.subplot(1,3,2)
plt.imshow(yuv)
plt.title("Imagen YUV")

plt.subplot(1,3,3)
plt.imshow(rgb_recuperado)
plt.title("YUV -> RGB")

plt.show()
```
