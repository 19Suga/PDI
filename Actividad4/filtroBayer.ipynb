```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

imagen = cv2.imread("colores.jpg", 0)  

# aplicar filtros Bayer
bayer_BG = cv2.cvtColor(imagen, cv2.COLOR_BAYER_BG2BGR)
bayer_GB = cv2.cvtColor(imagen, cv2.COLOR_BAYER_GB2BGR)
bayer_RG = cv2.cvtColor(imagen, cv2.COLOR_BAYER_RG2BGR)
bayer_GR = cv2.cvtColor(imagen, cv2.COLOR_BAYER_GR2BGR)

plt.figure(figsize=(10,8))

plt.subplot(2,2,1)
plt.imshow(cv2.cvtColor(bayer_BG, cv2.COLOR_BGR2RGB))
plt.title("Bayer BG")

plt.subplot(2,2,2)
plt.imshow(cv2.cvtColor(bayer_GB, cv2.COLOR_BGR2RGB))
plt.title("Bayer GB")

plt.subplot(2,2,3)
plt.imshow(cv2.cvtColor(bayer_RG, cv2.COLOR_BGR2RGB))
plt.title("Bayer RG")

plt.subplot(2,2,4)
plt.imshow(cv2.cvtColor(bayer_GR, cv2.COLOR_BGR2RGB))
plt.title("Bayer GR")

plt.show()
```
