ploteo por canales

```python
import cv2
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = [10, 4]

img = cv2.imread('colores.jpg', cv2.IMREAD_COLOR)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.subplot(141)
plt.title("Imagen original")
plt.imshow(img_rgb)
plt.axis("off")

plt.subplot(142)
plt.title("Canal Rojo")
plt.imshow(img[:,:,2], cmap='gray')
plt.axis("off")

plt.subplot(143)
plt.title("Canal Verde")
plt.imshow(img[:,:,1], cmap='gray')
plt.axis("off")

plt.subplot(144)
plt.title("Canal Azul")
plt.imshow(img[:,:,0], cmap='gray')
plt.axis("off")

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np  

plt.rcParams['figure.figsize'] = [10, 5]

img = cv2.imread('colores.jpg', cv2.IMREAD_COLOR)
imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Extraer canales desde imgRGB
r = imgRGB[:,:,0]
g = imgRGB[:,:,1]
b = imgRGB[:,:,2]

# Crear imagen por canal
r_img = np.zeros_like(imgRGB)
r_img[:,:,0] = r

g_img = np.zeros_like(imgRGB)
g_img[:,:,1] = g

b_img = np.zeros_like(imgRGB)
b_img[:,:,2] = b

plt.figure(figsize=(10,5))

plt.subplot(141)
plt.title("Imagen Original")
plt.imshow(imgRGB)
plt.axis("off")

plt.subplot(142)
plt.title("Canal Rojo")
plt.imshow(r_img)
plt.axis("off")

plt.subplot(143)
plt.title("Canal Verde")
plt.imshow(g_img)
plt.axis("off")

plt.subplot(144)
plt.title("Canal Azul")
plt.imshow(b_img)
plt.axis("off")

plt.show()


plt.show()
```
