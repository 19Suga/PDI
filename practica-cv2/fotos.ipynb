2-  Mostrar imagenes en formato BGR usando Matplotlib, convertida a RGB

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['figure.figsize'] = [5, 5]
img1 = cv2.imread('img1.jpeg')
img2 = cv2.imread('img2.jpeg')
img3 = cv2.imread('img3.jpeg')
img1RGB=cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
img2RGB=cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
img3RGB=cv2.cvtColor(img3, cv2.COLOR_BGR2RGB)

plt.imshow(img1RGB)
plt.show()
plt.imshow(img2RGB)
plt.show()
plt.imshow(img3RGB)
plt.show()
```

3- Convertir imagen BGR a escala de grises, utilizando cv2.cvtColor

```python
import cv2
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = [8, 4]

img1 = cv2.imread('img1.jpeg', cv2.IMREAD_COLOR)
img1_rgb = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)

img2 = cv2.imread('img2.jpeg', cv2.IMREAD_COLOR)
img2_rgb = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)

img3 = cv2.imread('img3.jpeg', cv2.IMREAD_COLOR)
img3_rgb = cv2.cvtColor(img3, cv2.COLOR_BGR2RGB)

# Visualizar la imagen1 original y sus canales de color
plt.subplot(141)
plt.title("Imagen original")
plt.imshow(img1_rgb)
plt.axis("off")

plt.subplot(142)
plt.title("Canal Rojo")
plt.imshow(img1[:,:,2], cmap='gray')
plt.axis("off")

plt.subplot(143)
plt.title("Canal Verde")
plt.imshow(img1[:,:,1], cmap='gray')
plt.axis("off")

plt.subplot(144)
plt.title("Canal Azul")
plt.imshow(img1[:,:,0], cmap='gray')
plt.axis("off")

plt.show()

# Visualizar la imagen2 original y sus canales de color
plt.subplot(141)
plt.title("Imagen original")
plt.imshow(img2_rgb)
plt.axis("off")

plt.subplot(142)
plt.title("Canal Rojo")
plt.imshow(img2_rgb[:,:,2], cmap='gray')
plt.axis("off")

plt.subplot(143)
plt.title("Canal Verde")
plt.imshow(img2_rgb[:,:,1], cmap='gray')
plt.axis("off")

plt.subplot(144)
plt.title("Canal Azul")
plt.imshow(img2_rgb[:,:,0], cmap='gray')
plt.axis("off")

plt.show()

# Visualizar la imagen3 original y sus canales de color
plt.subplot(141)
plt.title("Imagen original")
plt.imshow(img3_rgb)
plt.axis("off")

plt.subplot(142)
plt.title("Canal Rojo")
plt.imshow(img3_rgb [:,:,2], cmap='gray')
plt.axis("off")

plt.subplot(143)
plt.title("Canal Verde")
plt.imshow(img3_rgb [:,:,1], cmap='gray')
plt.axis("off")

plt.subplot(144)
plt.title("Canal Azul")
plt.imshow(img3_rgb [:,:,0], cmap='gray')
plt.axis("off")

plt.show()
```

4- Guardar la imagen convertida en el ordenador

```python
#GUARDAR IMAGEN CON OPENCV
cv2.imwrite('imagen_gris1.jpg', img1_rgb[:,:,2])
cv2.imwrite('imagen_gris2.jpg', img2_rgb[:,:,2])
cv2.imwrite('imagen_gris3.jpg', img3_rgb[:,:,2])
```

5- Guardar la imagen convertida en BGR en el ordenador

```python
plt.imsave('img_BGR1.jpg', cv2.cvtColor(img1_rgb, cv2.COLOR_BGR2RGB))
plt.imsave('img_BGR2.jpg', cv2.cvtColor(img2_rgb, cv2.COLOR_BGR2RGB))
plt.imsave('img_BGR3.jpg', cv2.cvtColor(img3_rgb, cv2.COLOR_BGR2RGB))
```

6- Funcion para mostrar imagenes 

```python
import cv2
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = [5, 5]

# funcion
def mostrar_imagen(ruta):
    img = cv2.imread(ruta)                 
    imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    plt.imshow(imgRGB)
    plt.axis('off')                  
    plt.show()

# llamada de funcion
mostrar_imagen('img1.jpeg')
mostrar_imagen('img2.jpeg')
mostrar_imagen('img3.jpeg')
```

7- Mostrar en diferentes canales en escala de grises, en formato de color y color falso 

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

img1 = cv2.imread('img1.jpeg')
img1RGB = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
imgGray1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)

R = img1RGB[:,:,0]
G = img1RGB[:,:,1]
B = img1RGB[:,:,2]

plt.figure(figsize=(18,6))

plt.subplot(1,6,1)
plt.imshow(img1RGB)
plt.title("Color")
plt.axis('off')

plt.subplot(1,6,2)
plt.imshow(imgGray1, cmap='gray')
plt.title("Grises")
plt.axis('off')

plt.subplot(1,6,3)
plt.imshow(imgGray1, cmap='jet')
plt.title("Color Falso")
plt.axis('off')

plt.subplot(1,6,4)
plt.imshow(R, cmap='gray')
plt.title("Canal R")
plt.axis('off')

plt.subplot(1,6,5)
plt.imshow(G, cmap='gray')
plt.title("Canal G")
plt.axis('off')

plt.subplot(1,6,6)
plt.imshow(B, cmap='gray')
plt.title("Canal B")
plt.axis('off')

plt.subplots_adjust(wspace=0.1, hspace=0.1)

plt.show()
```
