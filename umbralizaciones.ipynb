umbralizacion binaria y binaria inversa  (entrara al parcial)

```python
import cv2 
import matplotlib.pyplot as plt
imagen = cv2.imread('colores.jpg',0)#lee en escala de grises 
# umbralizacion binaria
_, umbral_binario = cv2.threshold(imagen, 100, 255, cv2.THRESH_BINARY)
# umbralizacion binaria inversa
_, umbral_binario_inv = cv2.threshold(imagen, 127, 255, cv2.THRESH_BINARY_INV)

plt.figure(figsize=(10,5))
plt.subplot(1,3,1)
plt.imshow(imagen, cmap='gray')
plt.title('Imagen Original')

plt.subplot(1,3,2)
plt.imshow(umbral_binario, cmap='gray')
plt.title('Imagen umbralizada binaria')

plt.subplot(1,3,3)
plt.imshow(umbral_binario_inv, cmap='gray')
plt.title('Imagen umbralizada binaria inversa')

plt.show()



```
