¿Qué es el Efecto Moiré?
El efecto moiré es un patrón de interferencia visual que se crea cuando se superponen dos rejillas, rejillas de puntos o patrones de líneas que tienen frecuencias, ángulos o tamaños ligeramente diferentes. El resultado es la aparición de un tercer patrón "fantasma" de mayor tamaño que no existe en ninguna de las dos imágenes originales.

El Fundamento Físico: Interferencia
Desde el punto de vista de la física y la óptica, el moiré ocurre debido a la superposición de estructuras periódicas. Cuando las líneas de un patrón coinciden con los espacios del otro, la luz se bloquea o se permite pasar de forma desigual, creando zonas de mayor y menor densidad visual.Matemáticamente, si tenemos dos patrones con frecuencias espaciales $f_1$ y $f_2$, la frecuencia del patrón moiré resultante ($f_m$) suele ser la diferencia entre ambas

El Moiré en el Mundo Digital (Aliasing)
En la fotografía y el desarrollo visual, el moiré suele considerarse un artefacto no deseado. Ocurre cuando la textura de un objeto (como el tejido de una camisa o una rejilla metálica) es más fina que la resolución del sensor de la cámara o la rejilla de píxeles de una pantalla.
Muestreo insuficiente: El sensor intenta "leer" un patrón que es más denso que sus propios fotodiodos, lo que genera una interpretación errónea de los datos.
Filtros OLPF: Muchas cámaras incluyen un filtro de paso bajo óptico (Optical Low Pass Filter) para suavizar ligeramente la imagen y evitar que estas frecuencias altas causen moiré.

Aplicaciones y Contextos
Aunque suele ser un error, el moiré tiene usos específicos:
Seguridad: Se utiliza en billetes y documentos oficiales como medida antifalsificación, ya que el patrón cambia o se vuelve ilegible al ser escaneado.
Metrología: Para detectar deformaciones minúsculas en materiales; si la superficie se dobla, el patrón moiré cambia drásticamente, revelando el movimiento.
Arte Op (Óptico): Artistas visuales lo utilizan para crear sensaciones de movimiento dinámico en superficies estáticas.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('moire.webp', cv2.IMREAD_GRAYSCALE)

img_reconstruida = cv2.GaussianBlur(img, (0, 0), sigmaX=8, sigmaY=8)

img_reconstruida = cv2.normalize(img_reconstruida, None, 0, 255, cv2.NORM_MINMAX)

img_ecualizada = cv2.equalizeHist(img_reconstruida)

plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(img, cmap='gray')
plt.title('Imagen original con moire')
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(img_reconstruida, cmap='gray')
plt.title('Reconstruccion suavizada')
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(img_ecualizada, cmap='gray')
plt.title('Reconstruccion mejorada')
plt.axis('off')

plt.tight_layout()
plt.show()
```

¿Qué es el aliasing?
El aliasing es el efecto que se produce cuando una señal continua se digitaliza con una frecuencia de muestreo menor a la necesaria, generando distorsión y pérdida de información en la representación digital.

¿En qué tipo de imágenes aparece?
El aliasing aparece principalmente en imágenes con:

1. Líneas muy finas
Rejillas
Rayas paralelas
Texturas repetitivas
2. Patrones de alta frecuencia
Telas
Pantallas digitales
Códigos de barras
3. Bordes diagonales o curvos
Se ven como “dientes de sierra”
4. Imágenes reducidas de tamaño sin filtrado previo
Downsampling sin suavizado

El aliasing es un fenómeno que ocurre cuando una imagen es muestreada con una resolución insuficiente, provocando la aparición de distorsiones visuales como bordes irregulares y patrones falsos. Este efecto se produce debido a la incapacidad del sistema digital para representar adecuadamente las altas frecuencias presentes en la imagen. El aliasing es común en imágenes con líneas finas, texturas repetitivas y bordes diagonales. Para evitarlo, se emplean técnicas como el filtrado pasa-bajas, el aumento de la resolución de muestreo y métodos de anti-aliasing.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('aliasing.jpg', cv2.IMREAD_GRAYSCALE)
#Gaussiano
img_gauss = cv2.GaussianBlur(img, (7, 7), 0)
#Filtro de media
img_media = cv2.blur(img, (5, 5))
#bilateral
img_bilateral = cv2.bilateralFilter(img, 9, 75, 75)

plt.figure(figsize=(16, 10))
plt.subplot(2, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Imagen original')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(img_gauss, cmap='gray')
plt.title('Filtro Gaussiano')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(img_media, cmap='gray')
plt.title('Filtro de Media')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(img_bilateral, cmap='gray')
plt.title('Filtro Bilateral')
plt.axis('off')

plt.tight_layout()
plt.show()
```
